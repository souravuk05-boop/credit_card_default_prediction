### Understand the important columns
- ID -> Customer record identifier
- LIMIT_BAL -> Credit limit in New Taiwan dollars
- SEX	-> Encoded sex category
- EDUCATION -> Education category
- MARRIAGE -> Marital status category
- AGE -> Customer age
- PAY_0 to PAY_6 -> Historical repayment status codes
- BILL_AMT1 to BILL_AMT6 -> Previous monthly bill amounts
- PAY_AMT1 to PAY_AMT6 -> Previous monthly payment amounts
- default.payment.next.month -> Target: 1 = Default, 0 = No default
